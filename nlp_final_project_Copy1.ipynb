```python
# Step 1 – Dataset

reviews = [
    "This movie was absolutely fantastic",
    "I loved every minute of this film",
    "The acting was brilliant and emotional",
    "Amazing movie with excellent performances",

    "This movie was terrible",
    "I hated every minute of it",
    "The acting was boring and disappointing",
    "Worst movie I have ever watched"
]

labels = [
    1, 1, 1, 1,
    0, 0, 0, 0
]
```

```python
# Step 2 – TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    ngram_range=(1, 2)
)

X_tfidf = tfidf.fit_transform(reviews)

print("TF-IDF shape:", X_tfidf.shape)
```

```python
# Step 3 - Train Sentiment Classifier
from sklearn.linear_model import LogisticRegression

classifier = LogisticRegression(
    random_state=42
)

classifier.fit(
    X_tfidf,
    labels
)
```

```python
# Step 4 - Test
new_reviews = [
    "The movie was amazing and brilliant",
    "This was a horrible and boring movie",
    "I hated every minute of it",
    "The acting was boring and disappointing"
]

X_new = tfidf.transform(new_reviews)
predictions = classifier.predict(X_new)

for review, prediction in zip(
    new_reviews,
    predictions
):
    sentiment = (
        "POSITIVE"
        if prediction == 1
        else "NEGATIVE"
    )
    print(review)
    print("Sentiment:", sentiment)
    print(type(sentiment))
    print()

# Text
# ↓
# TF-IDF
# ↓
# ML model
# ↓
# Sentiment
```

```python
# Step 5 – Semantic Similarity
# Now provide this same reviews to Sentence Transformer.

from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

review_embeddings = embedding_model.encode(
    reviews,
    normalize_embeddings=True
)

print(review_embeddings)
```

```python
# Step 6 – User Review

user_review = [
    "The film was wonderful and the acting was excellent."
]

user_embedding = embedding_model.encode(
    user_review,
    normalize_embeddings=True
)

print(user_embedding)
```

```python
# Step 7 - Cosine Similarity

from sklearn.metrics.pairwise import cosine_similarity

similarities = cosine_similarity(
    user_embedding,
    review_embeddings
)[0]

print(f"Similarities:{similarities}")

for review, score, label in zip(
    reviews,
    similarities,
    labels
):
    print(
        f"{score:.4f} | "
        f"{'POSITIVE' if label == 1 else 'NEGATIVE'} | "
        f"{review}"
    )

# User Review
# Embedding
# Cosine Similarity
# Most Similar Reviews
```

```python
# step-8 - Find top 3 Simiar Reviews

import numpy as np

top_k = 3

top_indices = np.argsort(
    similarities
)[::-1][:top_k]
print(np.argsort(
    similarities
))
print(top_indices)
print("Most Similar reviews:\n")

for idx in top_indices:
    print(
        f"Score: {similarities[idx]:.4f}"
    )

    print(
        f"Sentiment: "
        f"{'POSITIVE' if labels[idx] == 1 else 'NEGATIVE'}"
    )

    print(
        f"Review: {reviews[idx]}"
    )

    print("-" * 60)

# This introduces:
# Semantic Search
# without even needing a vector database yet.
```

```python
# Step 9 — Combine Sentiment + Similarity

# Now final function:

def analyze_review(review):

    # ----------------------------
    # 1. Sentiment
    # ----------------------------

    tfidf_vector = tfidf.transform([review])

    prediction = classifier.predict(
        tfidf_vector
    )[0]

    sentiment = (
        "POSITIVE"
        if prediction == 1
        else "NEGATIVE"
    )


    # ----------------------------
    # 2. Embedding
    # ----------------------------

    embedding = embedding_model.encode(
        [review],
        normalize_embeddings=True
    )


    # ----------------------------
    # 3. Similarity
    # ----------------------------

    similarities = cosine_similarity(
        embedding,
        review_embeddings
    )[0]


    # ----------------------------
    # 4. Top Similar Reviews
    # ----------------------------

    top_indices = np.argsort(
        similarities
    )[::-1][:3]
    print("=" * 70)
    print("REVIEW")
    print(review)

    print("\nSENTIMENT")
    print(sentiment)

    print("\nSIMILAR REVIEWS")

    for idx in top_indices:

        print(
            f"\nScore: {similarities[idx]:.4f}"
        )

        print(
            reviews[idx]
        )

    return {
        "review": review,
        "sentiment": sentiment,
        "similarities": similarities}
```

```python
# Test
result = analyze_review(
"The film was delightful, with captivating storytelling and performances that felt truly authentic."
)
```
